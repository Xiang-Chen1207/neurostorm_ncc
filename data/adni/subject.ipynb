```python
import re
import os

# 定义文件路径
input_file_path = '/home/chenx/code/neurostorm_ncc/data/adni/adni_ad_mni_test.txt'
# 定义输出文件路径 (为了防止数据丢失，建议保存为新文件)
output_file_path = '/home/chenx/code/neurostorm_ncc/data/adni/adni_ad_mni_test.txt'

def filter_unique_subjects(input_path, output_path):
    # 用于存储已经出现过的被试ID
    seen_subjects = set()
    # 用于存储过滤后的文件路径
    unique_lines = []

    # 正则表达式：匹配 sub- 后面跟着一串字符，直到遇到下划线 _ 为止
    # 针对示例：sub-116S6100_ses... 会提取出 sub-116S6100
    pattern = re.compile(r'(sub-[^_]+)')

    try:
        with open(input_path, 'r') as f:
            lines = f.readlines()

        print(f"原始文件共 {len(lines)} 行，正在处理...")

        for line in lines:
            line = line.strip()
            if not line:
                continue # 跳过空行

            # 在当前行中查找 sub-id
            match = pattern.search(line)
            
            if match:
                subject_id = match.group(1)
                
                # 如果这个被试ID不在集合中，说明是第一次遇到
                if subject_id not in seen_subjects:
                    seen_subjects.add(subject_id)
                    unique_lines.append(line)
            else:
                # 如果某行没有匹配到 sub-id，可以选择打印警告或保留/丢弃
                # 这里默认如果不包含被试信息则不保留，或者你可以根据需求修改
                print(f"Warning: 无法提取被试ID: {line}")

        # 将结果写入新文件
        with open(output_path, 'w') as f:
            for line in unique_lines:
                f.write(line + '\n')

        print(f"处理完成！")
        print(f"保留了 {len(unique_lines)} 个唯一被试的文件路径。")
        print(f"结果已保存至: {output_path}")

    except FileNotFoundError:
        print(f"错误: 找不到文件 {input_path}")
    except Exception as e:
        print(f"发生错误: {e}")

if __name__ == "__main__":
    filter_unique_subjects(input_file_path, output_file_path)
```
