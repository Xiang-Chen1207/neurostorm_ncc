```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# 读取 CSV
df = pd.read_csv("/home/chenx/code/neurostorm_ncc/data/abide/abide.csv")

# 保存原始年龄
df["AGE_AT_SCAN_raw"] = df["AGE_AT_SCAN"]

# 标准化 AGE_AT_SCAN
scaler = StandardScaler()
df["AGE_AT_SCAN"] = scaler.fit_transform(df[["AGE_AT_SCAN"]]).round(2)

# 将 AGE_AT_SCAN_raw 移动到最后一列
cols = list(df.columns)
cols.remove("AGE_AT_SCAN_raw")
cols.append("AGE_AT_SCAN_raw")
df = df[cols]

# 保存文件
df.to_csv("/home/chenx/code/neurostorm_ncc/data/abide/abide_scaled.csv", index=False)

print(df.head())
```
